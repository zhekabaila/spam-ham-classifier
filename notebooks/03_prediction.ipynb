# 03 - Prediksi Pesan

Gunakan model terlatih untuk memprediksi apakah pesan termasuk spam atau ham.

```python
import sys
sys.path.append('../')
from src.predictor import SpamHamPredictor

predictor = SpamHamPredictor(model_path='../models/spam_ham_model.pkl')
print("✅ Model berhasil dimuat!")
```

```python
teks_input = "Secara alami tak tertahankan identitas perusahaan Anda sangat sulit untuk mengingat kembali perusaha"

hasil = predictor.predict(teks_input)
predictor.display_result(hasil)
```

```python
daftar_pesan = [
    "Selamat! Anda memenangkan hadiah undian senilai Rp 50 juta. Hubungi kami segera.",
    "Hai, apa kabar? Besok kita jadi ketemuan jam 3 sore kan?",
    "PROMO TERBATAS! Diskon 90% hanya hari ini. Klik link berikut untuk klaim.",
    "Laporan keuangan bulan ini sudah saya kirim lewat email.",
    "Klik di sini untuk verifikasi akun Anda sebelum diblokir permanen.",
]

print(f"{'No':<4} {'Prediksi':<10} {'Keyakinan':<12} {'Pesan'}")
print("-" * 75)

for i, pesan in enumerate(daftar_pesan, 1):
    hasil = predictor.predict(pesan)
    label = hasil['prediksi'].upper()
    emoji = "🚫" if hasil['prediksi'] == 'spam' else "✅"
    print(f"{i:<4} {emoji} {label:<8} {hasil['keyakinan']:>8.1f}%    {pesan[:45]}...")
```

```python
def cek_pesan():
    """Fungsi interaktif untuk mengecek pesan."""
    print("=" * 55)
    print("  SPAM / HAM DETECTOR")
    print("=" * 55)
    while True:
        teks = input("\nMasukkan pesan (atau ketik 'keluar' untuk berhenti):\n> ").strip()
        if teks.lower() == 'keluar':
            print("Program selesai.")
            break
        if not teks:
            print("⚠️  Pesan tidak boleh kosong.")
            continue
        hasil = predictor.predict(teks)
        predictor.display_result(hasil)

cek_pesan()
```
